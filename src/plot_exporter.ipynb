# Plot Exporter Notebook

This notebook was used to simply export the plots from `streamlit_app.py`.

```python
from streamlit_app import *
import seaborn as sns
```

```python
# get the dataframe and all figures
cph_df = get_df()

histogram_map_np = get_2d_hist_np(cph_df, 100)
primary_restaurant_top20 = get_primary_restaurant_top20(cph_df)
quadtree_fig = get_quadtree(cph_df)
vegetarian_map = get_vegetarian_bar(cph_df)
rating_vs_price_boxplot = get_rating_vs_price_box(cph_df)
```

```python
vegetarian_map.update_layout(template='simple_white', width=800)
vegetarian_map.update_yaxes(showgrid=False)
# vegetarian_map.write_image(f"plots/vegetarian_map.svg", format="svg", width=800)
```
